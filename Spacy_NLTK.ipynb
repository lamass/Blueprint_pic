```python
# use quantra_py kernel
import spacy 
```

Sentence tokinization

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp('The silent transmission of spirit. It happens when spirit meets itself through a barrier.')
for sentence in doc.sents:
    print(sentence)
```

Word tokinization

```python
for s in doc.sents:
    for word in s:
        print(word)
```

NLTK

```python
# import nltk
nltk.download('punkt')
from nltk.tokenize import  sent_tokenize  
```

```python
sent_tokenize('The silent transmission of spirit. It happens when spirit meets itself through a barrier.')
```

```python
from nltk.tokenize import word_tokenize
word_tokenize('The silent transmission of spirit. It happens when spirit meets itself through a barrier.')
```
